```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
from sklearn.linear_model import Ridge,Lasso
ridge=Ridge(alpha=1.0)
ridge.fit(x_train,y_train)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score,root_mean_squared_error
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

```python
root_mean_squared_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python

def evaluate(y_test, prediction):
    mae = mean_absolute_error(y_test, prediction)
    mse = mean_squared_error(y_test, prediction)
    rmse = root_mean_squared_error(y_test,prediction)
    r2 = r2_score(y_test, prediction)

    return mae, mse, rmse, r2
evaluate(y_test,y_pred)
evaluate(y_test,y_pred_ridge)
evaluate(y_test,y_pred_lasso)
```

```python
print("Linear:", evaluate(y_test, y_pred))
print("Ridge:", evaluate(y_test, y_pred_ridge))
print("Lasso:", evaluate(y_test, y_pred_lasso))
```

Linear Regression performed best overall, achieving the lowest MSE and RMSE and the highest R² of approximately 0.9884. Ridge produced almost identical results, indicating that regularization provided little additional benefit, while Lasso performed worse because its stronger coefficient shrinkage increased prediction error.
